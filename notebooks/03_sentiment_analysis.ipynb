```python
# Score sentiment with a Twitter-RoBERTa model in GPU batches.
import gc
import numpy as np
import torch
import torch.nn.functional as F
from tqdm.auto import tqdm
from transformers import AutoModelForSequenceClassification, AutoTokenizer

SENT_MODEL = "cardiffnlp/twitter-roberta-base-sentiment-latest"
# Use GPU for transformer inference when Colab provides one.
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# Load the Twitter-specific sentiment tokenizer and classifier.
tokenizer = AutoTokenizer.from_pretrained(SENT_MODEL)
model = AutoModelForSequenceClassification.from_pretrained(
    SENT_MODEL,
    dtype=torch.float16 if device.type == "cuda" else torch.float32
).to(device).eval()

# Score sentiment on raw tweet text, not cleaned_text, so emotional cues are preserved.
texts = df["rawContent"].fillna("").astype(str).tolist()
# Preallocate arrays so batch results can be written by row position.
sentiment = np.full(len(df), np.nan)
sent_neg = np.full(len(df), np.nan)
sent_neu = np.full(len(df), np.nan)
sent_pos = np.full(len(df), np.nan)

# Large batches maximize A100 throughput; reduce this if Colab reports CUDA OOM.
batch_size = 4096

for start in tqdm(range(0, len(texts), batch_size), desc="Sentiment scoring"):
    # Tokenize and score one batch at a time to avoid loading all tensors onto GPU.
    end = min(start + batch_size, len(texts))
    batch = texts[start:end]
    enc = tokenizer(
        batch,
        return_tensors="pt",
        truncation=True,
        max_length=128,
        padding=True,
        pad_to_multiple_of=8
    ).to(device)
    with torch.inference_mode(), torch.autocast(device_type="cuda", dtype=torch.float16, enabled=(device.type == "cuda")):
        # Run inference without gradients and with mixed precision on CUDA.
        out = model(**enc)
    # Convert class probabilities into a continuous sentiment score from -1 to +1.
    probs = F.softmax(out.logits.float(), dim=1).cpu().numpy()
    sentiment[start:end] = -probs[:, 0] + probs[:, 2]
    sent_neg[start:end] = probs[:, 0]
    sent_neu[start:end] = probs[:, 1]
    sent_pos[start:end] = probs[:, 2]
    del enc, out, probs
    if start % (batch_size * 200) == 0 and start > 0:
        gc.collect()
        if torch.cuda.is_available():
            torch.cuda.empty_cache()

# Attach the continuous score and class probabilities to the tweet table.
df["sentiment"] = sentiment
df["sent_neg"] = sent_neg
df["sent_neu"] = sent_neu
df["sent_pos"] = sent_pos

# Save the scored corpus for state-level aggregation.
score_path = INTERIM_DIR / "tweets_with_topics_scores.parquet"
df.to_parquet(score_path, index=False)

print("Saved:", score_path)
display(df[["state", "topic", "sentiment", "rawContent"]].head())
```

```python
# Score stance with an NLI model, but keep it as a diagnostic variable because validation failed.
import gc
import numpy as np
import torch
import torch.nn.functional as F
from tqdm.auto import tqdm
from transformers import AutoModelForSequenceClassification, AutoTokenizer

STANCE_MODEL = "MoritzLaurer/DeBERTa-v3-base-mnli-fever-anli"
# Reuse GPU inference for the diagnostic stance sample when available.
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# Load the NLI model used for conservative/progressive diagnostic scoring.
stance_tokenizer = AutoTokenizer.from_pretrained(STANCE_MODEL)
stance_model = AutoModelForSequenceClassification.from_pretrained(
    STANCE_MODEL,
    dtype=torch.float16 if device.type == "cuda" else torch.float32
).to(device).eval()

# NLI hypotheses define the two political-direction scores to compare.
cons_hyp = "This text expresses conservative right-wing political views."
prog_hyp = "This text expresses progressive left-wing political views."

# Score only a reproducible sample because stance validation failed for final modelling.
STANCE_SAMPLE_SIZE = 500_000
stance_sample_size = min(STANCE_SAMPLE_SIZE, len(df))
df_stance = df.sample(n=stance_sample_size, random_state=42).reset_index(drop=True)
stance_texts = df_stance["rawContent"].fillna("").astype(str).tolist()

# Preallocate diagnostic stance outputs for the sampled tweets.
stance = np.full(len(df_stance), np.nan)
stance_conservative = np.full(len(df_stance), np.nan)
stance_progressive = np.full(len(df_stance), np.nan)

# NLI uses paired text/hypothesis inputs, so the stance batch is smaller than sentiment.
batch_size = 1024

for start in tqdm(range(0, len(stance_texts), batch_size), desc="Stance scoring"):
    # Compare each tweet against conservative and progressive hypotheses separately.
    end = min(start + batch_size, len(stance_texts))
    batch = stance_texts[start:end]

    cons_enc = stance_tokenizer(
        batch, [cons_hyp] * len(batch),
        return_tensors="pt",
        truncation=True,
        max_length=128,
        padding=True,
        pad_to_multiple_of=8
    ).to(device)
    prog_enc = stance_tokenizer(
        batch, [prog_hyp] * len(batch),
        return_tensors="pt",
        truncation=True,
        max_length=128,
        padding=True,
        pad_to_multiple_of=8
    ).to(device)

    with torch.inference_mode(), torch.autocast(device_type="cuda", dtype=torch.float16, enabled=(device.type == "cuda")):
        # Run both NLI hypothesis batches in mixed precision when CUDA is available.
        cons_out = stance_model(**cons_enc)
        prog_out = stance_model(**prog_enc)

    # Entailment probability is the model's support for each political hypothesis.
    cons_scores = F.softmax(cons_out.logits.float(), dim=1).cpu().numpy()[:, 2]
    prog_scores = F.softmax(prog_out.logits.float(), dim=1).cpu().numpy()[:, 2]

    stance[start:end] = cons_scores - prog_scores
    stance_conservative[start:end] = cons_scores
    stance_progressive[start:end] = prog_scores

    del cons_enc, prog_enc, cons_out, prog_out, cons_scores, prog_scores
    if start % (batch_size * 200) == 0 and start > 0:
        gc.collect()
        if torch.cuda.is_available():
            torch.cuda.empty_cache()

# Save stance as diagnostic metadata only, not as a final SEM predictor.
df_stance["stance"] = stance
df_stance["stance_conservative"] = stance_conservative
df_stance["stance_progressive"] = stance_progressive

print("Stance was computed on a diagnostic sample only, because validation showed it was not reliable enough for modelling.")
print("Keep this output for inspection only, not as a final SEM predictor.")

# Keep the stance sample for validation and thesis transparency.
stance_path = INTERIM_DIR / "stance_diagnostic_sample.parquet"
df_stance.to_parquet(stance_path, index=False)

print("Saved:", stance_path)
display(df_stance[["state", "topic", "stance", "rawContent"]].head())
```
