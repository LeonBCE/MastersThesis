```python
# Define reusable helpers for state extraction and tweet cleaning.
import ast
import html
import json
import re
import pandas as pd

# State names and abbreviations are used to normalize user location strings.
US_STATES = {
    "Alabama": "AL", "Alaska": "AK", "Arizona": "AZ", "Arkansas": "AR", "California": "CA",
    "Colorado": "CO", "Connecticut": "CT", "Delaware": "DE", "Florida": "FL", "Georgia": "GA",
    "Hawaii": "HI", "Idaho": "ID", "Illinois": "IL", "Indiana": "IN", "Iowa": "IA",
    "Kansas": "KS", "Kentucky": "KY", "Louisiana": "LA", "Maine": "ME", "Maryland": "MD",
    "Massachusetts": "MA", "Michigan": "MI", "Minnesota": "MN", "Mississippi": "MS",
    "Missouri": "MO", "Montana": "MT", "Nebraska": "NE", "Nevada": "NV",
    "New Hampshire": "NH", "New Jersey": "NJ", "New Mexico": "NM", "New York": "NY",
    "North Carolina": "NC", "North Dakota": "ND", "Ohio": "OH", "Oklahoma": "OK",
    "Oregon": "OR", "Pennsylvania": "PA", "Rhode Island": "RI", "South Carolina": "SC",
    "South Dakota": "SD", "Tennessee": "TN", "Texas": "TX", "Utah": "UT",
    "Vermont": "VT", "Virginia": "VA", "Washington": "WA", "West Virginia": "WV",
    "Wisconsin": "WI", "Wyoming": "WY",
}

# Common city names help recover states from informal Twitter profile locations.
CITY_TO_STATE = {
    "nyc": "New York", "new york": "New York", "brooklyn": "New York", "manhattan": "New York",
    "queens": "New York", "bronx": "New York", "buffalo": "New York", "rochester": "New York",
    "los angeles": "California", "san francisco": "California", "san diego": "California",
    "sacramento": "California", "oakland": "California", "san jose": "California",
    "chicago": "Illinois", "houston": "Texas", "san antonio": "Texas", "dallas": "Texas",
    "austin": "Texas", "fort worth": "Texas", "el paso": "Texas", "phoenix": "Arizona",
    "tucson": "Arizona", "philadelphia": "Pennsylvania", "philly": "Pennsylvania",
    "pittsburgh": "Pennsylvania", "miami": "Florida", "tampa": "Florida", "orlando": "Florida",
    "jacksonville": "Florida", "atlanta": "Georgia", "savannah": "Georgia",
    "detroit": "Michigan", "grand rapids": "Michigan", "ann arbor": "Michigan",
    "seattle": "Washington", "spokane": "Washington", "portland": "Oregon",
    "denver": "Colorado", "boulder": "Colorado", "nashville": "Tennessee",
    "memphis": "Tennessee", "knoxville": "Tennessee", "boston": "Massachusetts",
    "cambridge": "Massachusetts", "baltimore": "Maryland", "milwaukee": "Wisconsin",
    "madison": "Wisconsin", "minneapolis": "Minnesota", "st paul": "Minnesota",
    "new orleans": "Louisiana", "baton rouge": "Louisiana", "las vegas": "Nevada",
    "vegas": "Nevada", "reno": "Nevada", "albuquerque": "New Mexico",
    "santa fe": "New Mexico", "kansas city": "Missouri", "st louis": "Missouri",
    "omaha": "Nebraska", "lincoln": "Nebraska", "louisville": "Kentucky",
    "lexington": "Kentucky", "richmond": "Virginia", "virginia beach": "Virginia",
    "birmingham": "Alabama", "montgomery": "Alabama", "salt lake city": "Utah",
    "des moines": "Iowa", "boise": "Idaho", "anchorage": "Alaska", "honolulu": "Hawaii",
    "providence": "Rhode Island", "newark": "New Jersey", "trenton": "New Jersey",
    "charleston": "South Carolina", "columbia": "South Carolina", "little rock": "Arkansas",
    "jackson": "Mississippi", "hartford": "Connecticut", "wilmington": "Delaware",
    "burlington": "Vermont", "fargo": "North Dakota", "sioux falls": "South Dakota",
    "billings": "Montana", "cheyenne": "Wyoming", "manchester": "New Hampshire",
    "bangor": "Maine", "wichita": "Kansas", "topeka": "Kansas",
}

# Candidate names and platform noise are stripped before topic modelling.
REMOVE_WORDS = {
    "amp", "rt", "retweet", "repost", "reposted", "follow", "following", "breaking",
    "official", "thread", "link", "click", "watch", "video", "photo",
    "trump", "donald", "biden", "joe", "kamala", "harris", "maga", "potus",
    "vance", "walz", "obama", "clinton", "hillary", "rfk", "kennedy",
    "nikki", "haley", "jill", "stein", "vivek", "ramaswamy",
    "pelosi", "tulsi", "gabbard", "melania", "desantis",
    "cheney", "czar", "kama",
}

def extract_location_from_user(user_value):
    # Parse the nested user object and return the profile location when present.
    if user_value is None or pd.isna(user_value):
        return None
    if isinstance(user_value, dict):
        return user_value.get("location")
    text_value = str(user_value)
    for parser in (json.loads, ast.literal_eval):
        try:
            parsed = parser(text_value)
            if isinstance(parsed, dict):
                return parsed.get("location")
        except Exception:
            pass
    match = re.search(r"'location':\s*'([^']+)'", text_value)
    return match.group(1) if match else None

def get_state(location):
    # Convert a free-text location string into a US state name, excluding DC.
    if location is None or pd.isna(location) or not isinstance(location, str):
        return None
    loc = location.lower().strip()
    if "district of columbia" in loc or "washington dc" in loc or "washington, dc" in loc:
        return None
    if re.search(r"[,\s]dc$", loc):
        return None
    for full, abbr in US_STATES.items():
        full_l = full.lower()
        abbr_l = abbr.lower()
        if full_l in loc:
            return full
        if loc == abbr_l:
            return full
        if re.search(rf",\s*{abbr_l}$", loc):
            return full
    for city, state in CITY_TO_STATE.items():
        if city in loc:
            return state
    return None

def clean_text(text):
    # Normalize tweet text for BERTopic while keeping issue words and hashtag text.
    if text is None or pd.isna(text):
        return ""
    text = html.unescape(str(text))
    text = re.sub(r"http\S+|www\S+", " ", text)
    text = re.sub(r"@\w+", " ", text)
    text = re.sub(r"#[A-Za-z0-9_]+", lambda m: " " + m.group(0)[1:] + " ", text)
    text = re.sub(r"[^a-zA-Z0-9\s]", " ", text)
    text = text.lower()
    words = [w for w in text.split() if w not in REMOVE_WORDS and len(w) > 2]
    text = " ".join(words)
    text = re.sub(r"\b(\w+)(?:\s+\1\b)+", r"\1", text)
    return re.sub(r"\s+", " ", text).strip()
```

```python
# Read parquet files in pandas, keep English state-mapped tweets, and save cleaned batches.
import gc
import shutil
from tqdm.auto import tqdm

# Recreate the batch folder so reruns do not mix old and new outputs.
batch_dir = INTERIM_DIR / "english_state_cleaned_batches"
if batch_dir.exists():
    shutil.rmtree(batch_dir)
batch_dir.mkdir(parents=True, exist_ok=True)

# Read only columns needed downstream to reduce memory pressure.
candidate_columns = [
    "id", "id_str", "date", "lang", "rawContent", "text", "location", "user",
    "replyCount", "retweetCount", "likeCount", "quoteCount", "viewCount"
]

# Track row counts at each filtering step for transparent reporting.
total_rows = 0
english_rows = 0
state_rows = 0
kept_rows = 0
batch_id = 0
# Accumulate cleaned frames until the parquet batch size threshold is reached.
buffer = []

for file_path in tqdm(files_to_process, desc="Processing parquet files"):
    # Process each parquet file independently so the full dataset can restart cleanly.
    try:
        with fs.open(file_path, "rb") as f:
            header_df = pd.read_parquet(f, columns=None)

        available_columns = [col for col in candidate_columns if col in header_df.columns]
        df = header_df[available_columns].copy()
        del header_df

        total_rows += len(df)

        # Preserve tweet IDs as strings so large numeric IDs are not rounded.
        for id_col in ["id", "id_str"]:
            if id_col in df.columns:
                df[id_col] = df[id_col].astype("string")

        # Keep English tweets only before doing location and text cleaning.
        if "lang" in df.columns:
            df = df[df["lang"].eq("en")].copy()
        english_rows += len(df)

        if "rawContent" not in df.columns and "text" in df.columns:
            df["rawContent"] = df["text"]

        if "location" not in df.columns:
            df["location"] = None

        df["location"] = df["location"].astype("object")

        # Fall back to user.location when the top-level location field is missing.
        missing_location = df["location"].isna() | df["location"].eq("")
        if "user" in df.columns:
            df.loc[missing_location, "location"] = df.loc[missing_location, "user"].apply(extract_location_from_user)

        # Keep only tweets that can be mapped to one of the 50 states.
        df["state"] = df["location"].apply(get_state)
        df = df[df["state"].notna()].copy()
        state_rows += len(df)

        # Clean text after state filtering so the expensive string work is smaller.
        df["rawContent"] = df["rawContent"].fillna("").astype(str)
        df["cleaned_text"] = df["rawContent"].apply(clean_text)
        df["word_count"] = df["cleaned_text"].str.split().str.len()
        df = df[df["word_count"] >= 3].copy()

        # Keep a compact schema for downstream topic, sentiment, and aggregation notebooks.
        keep_cols = [
            col for col in [
                "id", "id_str", "date", "state", "rawContent", "cleaned_text",
                "word_count", "replyCount", "retweetCount", "likeCount", "quoteCount", "viewCount"
            ]
            if col in df.columns
        ]

        # Write periodic parquet batches to avoid holding the full cleaned corpus in RAM.
        buffer.append(df[keep_cols])
        kept_rows += len(df)

        buffered_rows = sum(len(part) for part in buffer)
        if buffered_rows >= OUTPUT_BATCH_SIZE:
            batch_path = batch_dir / f"cleaned_batch_{batch_id:04d}.parquet"
            pd.concat(buffer, ignore_index=True).to_parquet(batch_path, index=False)
            buffer = []
            batch_id += 1

        del df
        gc.collect()

    except Exception as e:
        print(f"Skipped {file_path}: {e}")

# Flush any remaining rows after the final parquet file is processed.
if buffer:
    batch_path = batch_dir / f"cleaned_batch_{batch_id:04d}.parquet"
    pd.concat(buffer, ignore_index=True).to_parquet(batch_path, index=False)
    batch_id += 1

print(f"Raw rows scanned: {total_rows:,}")
print(f"English rows: {english_rows:,}")
print(f"English + state rows before text filter: {state_rows:,}")
print(f"Final kept rows: {kept_rows:,}")
print(f"Saved batches: {batch_id:,}")
print("Batch folder:", batch_dir)
```

```python
# Combine all cleaned parquet batches into one file for the next notebook.
# Combine all batch parquet files into the single cleaned dataset used downstream.
batch_files = sorted(batch_dir.glob("*.parquet"))
print(f"Batch files: {len(batch_files):,}")

df_clean = pd.concat([pd.read_parquet(path) for path in batch_files], ignore_index=True)
# Save the canonical cleaned tweet table for the rest of the pipeline.
clean_path = INTERIM_DIR / "english_state_cleaned.parquet"
df_clean.to_parquet(clean_path, index=False)

print(f"Final cleaned rows: {len(df_clean):,}")
print("Saved:", clean_path)
display(df_clean.head())
```
